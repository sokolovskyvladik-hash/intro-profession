# Поиск фильма через OMDb API

Ноутбук отправляет REST-запрос к OMDb API, разбирает JSON-ответ и показывает основные сведения о найденном фильме.


```python
import requests
from getpass import getpass

API_URL = "https://www.omdbapi.com/"
movie_title = input("Введите название фильма: ").strip()
api_key = getpass("Введите личный OMDb API-ключ: ")

if not movie_title:
    raise ValueError("Название фильма не должно быть пустым")
if not api_key:
    raise ValueError("API-ключ не должен быть пустым")

response = requests.get(
    API_URL,
    params={"apikey": api_key, "t": movie_title},
    timeout=15,
)
response.raise_for_status()
movie = response.json()

if movie.get("Response") == "True":
    print("Название:", movie.get("Title", "Не указано"))
    print("Год:", movie.get("Year", "Не указан"))
    print("Жанр:", movie.get("Genre", "Не указан"))
    print("Режиссёр:", movie.get("Director", "Не указан"))
else:
    print("Фильм не найден:", movie.get("Error", "Неизвестная ошибка"))
```
