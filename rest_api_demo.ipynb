# Поиск фильма через OMDb API

Ноутбук отправляет REST-запрос к OMDb API, разбирает JSON-ответ и показывает основные сведения о найденном фильме.


```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
API_URL = "https://www.omdbapi.com/"
api_key = os.getenv("OMDB_API_KEY")
movie_title = input("Введите название фильма: ").strip()

if not api_key:
    raise RuntimeError("Добавьте OMDB_API_KEY в локальный файл .env")
if not movie_title:
    raise ValueError("Название фильма не должно быть пустым")

response = requests.get(API_URL, params={"apikey": api_key, "t": movie_title}, timeout=15)
response.raise_for_status()
movie = response.json()

if movie.get("Response") == "True":
    for field, label in [("Title", "Название"), ("Year", "Год"), ("Genre", "Жанр"), ("Director", "Режиссёр"), ("Actors", "Актёры"), ("imdbRating", "Рейтинг IMDb")]:
        print(f"{label}: {movie.get(field, 'Не указано')}")
else:
    print("Ошибка OMDb:", movie.get("Error", "Неизвестная ошибка"))
```
